```python
# Mistake in our data sets:
# which i have to remove for the increases the quality of data

# Year has many non year values
# convert year object to int
# Price object to int
# kms_driven has kms with integers
# kms_driven object to int
# kms_driven has nan values
# fuel type has nan_vlue
# keep first 3 word of name
```

Cleaning

```python
backup=car.copy()
```

```python
# Year has many non year values
car=car[car['year'].str.isnumeric()]
```

```python
# convert year object to int
car['year']=car['year'].astype('Int32')
```

```python
car.info()
```

```python
# Remove ask for price from price column
car=car[car['Price']!="Ask For Price"]
```

```python
# Price has commas in its prices and is in object
car["Price"]=car["Price"].str.replace(',','').astype(int)
```

```python
# kms_driven has kms with integers
car['kms_driven']=car['kms_driven'].str.split(' ').str.get(0).str.replace(',','')
```

```python
# kms_driven has object values with kms at last.
car=car[car['kms_driven'].str.isnumeric()]
```

```python
# kms_driven object to int
car['kms_driven']=car['kms_driven'].astype('Int32')
```

```python
# fuel type has nan_vlue
car=car[~car['fuel_type'].isnull()]
```

```python
car['name']
```

```python
# keep first 3 word of name
car['name']=car['name'].str.split(' ').str.slice(0,3).str.join(' ')
```

```python
car=car.reset_index(drop=True)
```

```python
# check outliers
car=car[car['Price']<6e6].reset_index(drop=True)
```

Model

```python
x=car.drop(columns="Price")
y=car['Price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
```

```python
ohe=OneHotEncoder(handle_unknown='ignore')
ohe.fit(x[['name','company','fuel_type']])
```

```python
column_trans=make_column_transformer((OneHotEncoder(categories=ohe.categories_),['name','company','fuel_type']),remainder='passthrough')
```

```python
lr=LinearRegression()
```

```python
pipe=make_pipeline(column_trans,lr)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred=pipe.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

```python
scores=[]
for i in range(1000):
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=i)
    lr=LinearRegression()
    pipe=make_pipeline(column_trans,lr)
    pipe.fit(x_train,y_train)
    y_pred=pipe.predict(x_test)
    scores.append(r2_score(y_test,y_pred))
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
for i in range(1000):
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=np.argmax(scores))
    lr=LinearRegression()
    pipe=make_pipeline(column_trans,lr)
    pipe.fit(x_train,y_train)
    y_pred=pipe.predict(x_test)
    r2_score(y_test,y_pred)
```
